```python
n = int(input("Enter the limit: "))
a, b = 0, 1
print("Fibonacci Sequence:")
while a <= n:
    print(a, end=" ")
    a, b = b, a + b
```

```python
def rotate_list(lst, k):
    if not lst:
        return lst
    k = k % len(lst)
    return lst[-k:] + lst[:-k]
numbers = [1, 2, 3, 4, 5]
k = 2
result = rotate_list(numbers, k)
print("Original List:", numbers)
print("Rotated List:", result)
```
